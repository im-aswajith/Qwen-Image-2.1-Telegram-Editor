# ⚡ Qwen-Image 2.1 Telegram Editor — T4 Launcher

GPU target: **Google Colab NVIDIA T4**.

This notebook installs the repository runtime and launches the Telegram image-editing bot.


```python
# @title 1. Configure
import os, getpass

REPO_URL = "https://github.com/im-aswajith/Qwen-Image-2.1-Telegram-Editor.git"  # @param {type:"string"}
BRANCH = "main"  # @param {type:"string"}
TELEGRAM_TOKEN = getpass.getpass("Telegram bot token: ")
PROMPT = "Create a high-quality realistic edit while preserving the subject identity and important visual details."  # @param {type:"string"}

os.environ["TELEGRAM_TOKEN"] = TELEGRAM_TOKEN
os.environ["QWEN_PROMPT"] = PROMPT
print("Configuration loaded. Token is stored only in this runtime environment.")
```

```python
# @title 2. Verify T4
import subprocess, os, textwrap
print(subprocess.run(["nvidia-smi"], capture_output=True, text=True).stdout)
```

```python
# @title 3. Clone project
import os, subprocess
if not os.path.exists("/content/Qwen-Image-2.1-Telegram-Editor"):
    subprocess.run(["git", "clone", "--branch", BRANCH, REPO_URL, "/content/Qwen-Image-2.1-Telegram-Editor"], check=True)
   
```

```python
# @title 4. Launch
import os
path = "/content/Qwen-Image-2.1-Telegram-Editor/qwen-bot.py"
text = open(path, "r", encoding="utf-8").read()
text = text.replace('TELEGRAM_TOKEN = "YOUR_BOT_TOKEN"', 'TELEGRAM_TOKEN = os.environ["TELEGRAM_TOKEN"]')
text = text.replace('PROMPT = "YOUR-PROMPT"', 'PROMPT = os.environ.get("QWEN_PROMPT", "Create a high-quality realistic edit while preserving the subject identity and important visual details.")')
open(path, "w", encoding="utf-8").write(text)
os.chdir("/content/Qwen-Image-2.1-Telegram-Editor")
!python qwen-bot.py
```
